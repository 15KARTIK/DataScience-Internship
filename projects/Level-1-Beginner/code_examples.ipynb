```python
import seaborn as sns
import plotly.express as px
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load sample dataset
tips = sns.load_dataset("tips")
iris = px.data.iris()

print("✅ Setup complete")
```

```python
# Seaborn Line Plot
plt.figure(figsize=(8, 4))
sns.lineplot(x="size", y="total_bill", data=tips, errorbar=None)
plt.title("Seaborn - Line Plot")
plt.show()

# Plotly Line Plot
fig = px.line(tips.groupby("size")["total_bill"].mean().reset_index(),
              x="size", y="total_bill", title="Plotly - Line Plot")
fig.show()

print("✅ Line plot done")
```

```python
# Seaborn Scatter
plt.figure(figsize=(8, 4))
sns.scatterplot(x="total_bill", y="tip", hue="sex", data=tips)
plt.title("Seaborn - Scatter Plot")
plt.show()

# Plotly Scatter
fig = px.scatter(tips, x="total_bill", y="tip", color="sex", title="Plotly - Scatter Plot")
fig.show()

print("✅ Scatter plot done")
```

```python
# Seaborn Bar
plt.figure(figsize=(8, 4))
sns.barplot(x="day", y="total_bill", data=tips)
plt.title("Seaborn - Bar Chart")
plt.show()

# Plotly Bar
avg = tips.groupby("day")["total_bill"].mean().reset_index()
fig = px.bar(avg, x="day", y="total_bill", title="Plotly - Bar Chart")
fig.show()

print("✅ Bar chart done")
```

```python
# Seaborn Histogram
plt.figure(figsize=(8, 4))
sns.histplot(tips["total_bill"], bins=20, kde=True)
plt.title("Seaborn - Histogram")
plt.show()

# Plotly Histogram
fig = px.histogram(tips, x="total_bill", nbins=20, title="Plotly - Histogram")
fig.show()

print("✅ Histogram done")
```

```python
# Seaborn Box
plt.figure(figsize=(8, 4))
sns.boxplot(x="day", y="total_bill", data=tips)
plt.title("Seaborn - Box Plot")
plt.show()

# Plotly Box
fig = px.box(tips, x="day", y="total_bill", title="Plotly - Box Plot")
fig.show()

print("✅ Box plot done")
```

```python
# Seaborn doesn't have a native pie chart, so we use Matplotlib
plt.figure(figsize=(6, 6))
tips["day"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("Pie Chart - Tips by Day")
plt.ylabel("")
plt.show()

# Plotly Pie Chart
fig = px.pie(tips, names="day", title="Plotly - Pie Chart")
fig.show()

print("✅ Pie chart done")
```

```python
# Seaborn Count Plot
plt.figure(figsize=(8, 4))
sns.countplot(x="day", data=tips)
plt.title("Seaborn - Count Plot")
plt.show()

# Plotly Count (using histogram)
fig = px.histogram(tips, x="day", title="Plotly - Count Plot")
fig.show()

print("✅ Count plot done")
```

```python
# Seaborn Heatmap (correlation matrix)
plt.figure(figsize=(8, 6))
corr = tips[["total_bill", "tip", "size"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Seaborn - Heatmap")
plt.show()

# Plotly Heatmap
fig = px.imshow(corr, text_auto=True, color_continuous_scale="RdBu_r",
                title="Plotly - Heatmap")
fig.show()

print("✅ Heatmap done")
```

```python
# Seaborn Violin Plot
plt.figure(figsize=(8, 4))
sns.violinplot(x="day", y="total_bill", data=tips)
plt.title("Seaborn - Violin Plot")
plt.show()

# Plotly Violin Plot
fig = px.violin(tips, x="day", y="total_bill", box=True,
                title="Plotly - Violin Plot")
fig.show()

print("✅ Violin plot done")
```

```python
# Seaborn KDE Plot
plt.figure(figsize=(8, 4))
sns.kdeplot(data=tips, x="total_bill", hue="sex", fill=True)
plt.title("Seaborn - KDE Plot")
plt.show()

# Plotly KDE Plot (using histogram with marginal)
fig = px.histogram(tips, x="total_bill", color="sex",
                   marginal="box", title="Plotly - KDE-like Plot")
fig.show()

print("✅ KDE plot done")
```
